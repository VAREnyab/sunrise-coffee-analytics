```python
# Import libraries
from pathlib import Path
import pandas as pd
```

```python
# Locate the repository from either its root or the notebook directory.
project_root = Path.cwd().resolve()
while not (project_root / "pyproject.toml").is_file():
    if project_root == project_root.parent:
        raise FileNotFoundError("Open this notebook from inside the project repository.")
    project_root = project_root.parent

data_dir = project_root / "data"
data_dir
```

```python
data = {}

paths = {
    'dim_date':data_dir / "dim_date.csv",
    'dim_product':data_dir / "dim_product.csv",
    'dim_store':data_dir / "dim_store.csv",
    'fact_daily_sales':data_dir / "fact_daily_sales.csv",
    'fact_store_monthly_profitability':data_dir / "fact_store_monthly_profitability.csv"
}
```

```python
for csv_name, path in paths.items():
    data[csv_name] = pd.read_csv(path)
```

```python
# 1. Daily sales + store + date + product
daily_sales_enriched = (
    data["fact_daily_sales"]
    .merge(data["dim_store"], on="store_id", how="left")
    .merge(data["dim_date"], on="date_id", how="left")
    .merge(data["dim_product"], on="product_id", how="left")
)

daily_sales_enriched
```

```python
# 2. Monthly profitability + store + date
monthly_profitability_enriched = (
    data["fact_store_monthly_profitability"]
    .merge(data["dim_store"], on="store_id", how="left")
    .merge(data["dim_date"], on="date_id", how="left")
)

monthly_profitability_enriched
```
