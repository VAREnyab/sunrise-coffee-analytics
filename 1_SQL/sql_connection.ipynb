```python
# Import libraries
from pathlib import Path
import pandas as pd
import sqlite3
```

```python
# Locate the repository from either its root or the notebook directory.
project_root = Path.cwd().resolve()
while not (project_root / "pyproject.toml").is_file():
    if project_root == project_root.parent:
        raise FileNotFoundError("Open this notebook from inside the project repository.")
    project_root = project_root.parent

data_dir = project_root / "data"
data_dir

data = {}

paths = {
    'dim_date':data_dir / "dim_date.csv",
    'dim_product':data_dir / "dim_product.csv",
    'dim_store':data_dir / "dim_store.csv",
    'fact_daily_sales':data_dir / "fact_daily_sales.csv",
    'fact_store_monthly_profitability':data_dir / "fact_store_monthly_profitability.csv"
}

for csv_name, path in paths.items():
    data[csv_name] = pd.read_csv(path)
```

```python
# Create a SQLite database in your project root
db_path = project_root / "1_SQL" / "analytics.db"

conn = sqlite3.connect(db_path)

print(f"Connected to database: {db_path}")
```

```python
for table_name, df in data.items():
    df.to_sql(
        name=table_name,
        con=conn,
        if_exists="replace",
        index=False
    )

print("All CSV files have been loaded into SQLite!")
```

```python
# Reload the updated CSV
data["fact_store_monthly_profitability"] = pd.read_csv(
    data_dir / "fact_store_monthly_profitability.csv"
)

# Update the SQLite table
data["fact_store_monthly_profitability"].to_sql(
    name="fact_store_monthly_profitability",
    con=conn,
    if_exists="replace",
    index=False
)

print("Database table updated successfully!")
```
